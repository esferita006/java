```python
%%writefile Proyecto_Big_Brother.java
import java.util.Scanner;

public class Proyecto_Big_Brother {
    public static void main (String []args){

        //-------------------------------------------------------------------------------------------------------------------------------------
        //Ejercicio 1

        Scanner Scan1 = new Scanner(System.in);
        System.out.println("Ingrese la velocidad del sujeto vigilado ");
        int velocidad = Scan1.nextInt();

        while (velocidad <= 0){
            System.out.println("Dame una velocidad valida, es decir, mayor que cero");
            velocidad = Scan1.nextInt();
        }

        if (velocidad > 0 && velocidad <= 5){
            System.out.println("EL GRAN HERMANO ESTA VILANDOTE");
        } else if (velocidad > 5 && velocidad <= 10){
            System.out.println("Por favor, Reduce tu Velocidad :D ");
        } else if (velocidad > 10 && velocidad <= 15){
            System.out.println("Por tus recientes, comportamientos te indicamos que recogas un folleto de buen comportamiento");
            System.out.println("Se ha notificado a los trabajadores sobre tu comportamiento, OJO");
        } else{
            System.out.println("Las autoridades han sido alertadas sobre tu comportamiento, quedate donde estas");
        }
        System.out.println("Chequeo Terminado, Camarada");
        System.out.println("-------------------------------------");

        //-------------------------------------------------------------------------------------------------------------------------------------
        //Ejercicio 2

        Scanner Scan2 = new Scanner(System.in);
        Scanner Scan3 = new Scanner(System.in);
        System.out.println("Calificador 1,califique al sujeto");
        int calificacion1 = Scan2.nextInt();

        System.out.println("Calificador DOS, califique");
        int calificacion2 = Scan3.nextInt();

        if (calificacion1 <= 0 || calificacion2 <= 0){
            System.out.println("Tu ingreso de Hoy es : 0 Monedas");
            System.out.println("Esfuerzate mas la siguiente vez");
        } else if (calificacion1 == calificacion2){
            int Salario = calificacion1 + calificacion2;
            System.out.println("Tu salario de hoy es");
            System.out.println(Salario);
        } else if ((calificacion1 != calificacion2) && (calificacion1 < 2 || calificacion2 < 2)){
            System.out.println("Solo tendras DOS monedas Hoy");
        } else {
            int Salario = calificacion1 * calificacion2;
            System.out.println("Tu salario de hoy es");
            System.out.println(Salario);
        }

        System.out.println("-------------------------------------");
        //-------------------------------------------------------------------------------------------------------------------------------------
        //Ejercicio 3

        System.out.println("Hola Ciudadano, la hora de pagar el impuesto para la lucha Contrarevolucionaria ha llegado");
        System.out.println("Recuerda que cada mes son 100 pesos , mas los adeudos si es que tienes");

        System.out.println("Cuanto debe del mes anterior, sea honesto pls");
        Scanner ScanA = new Scanner(System.in);
        double LoDebido = ScanA.nextInt();

        System.out.println("Cuanto va a ingresar para este mes");
        Scanner Scan4 = new Scanner(System.in);
        double impuesto = Scan4.nextDouble();

        double impuestoReal = 100 + LoDebido;
        double LoDebo = impuestoReal - impuesto;
        double penalizacion;
        double NextTax;

        if (impuesto == impuestoReal){

            double suerte = Math.floor(Math.random()*6);

            if (suerte % 2 == 0){
                NextTax = 90;
                System.out.println("Usted debe un extra para el siguiente mes de");
                System.out.println(LoDebo);
                System.out.println("Su penalizacion por tanto sera de ");
                penalizacion = LoDebo/0.05;
                System.out.println(penalizacion);
                System.out.println("El siguiente pago TOTAL sera de  ");
                System.out.println(NextTax);
            }else{
                System.out.println("PARA OPERADOR: El siguiente Pago ha ganado 10 monedas");
                NextTax = 100;
                System.out.println("Usted debe un extra para el siguiente mes de");
                System.out.println(LoDebo);
                System.out.println("Su penalizacion por tanto sera de ");
                penalizacion = LoDebo/0.05;
                System.out.println(penalizacion);
                System.out.println("El siguiente pago TOTAL sera de  ");
                System.out.println(NextTax);
            }
        } else if (impuesto >= impuestoReal/2){
            penalizacion = LoDebo/20;
            NextTax = 100 + LoDebo + penalizacion;
            System.out.println("Para este mes le ha faltado este monto para completar su pago: ");
            System.out.println(LoDebo);
            System.out.println("Su penalizacion por tanto sera de ");
            System.out.println(penalizacion);
            System.out.println("El siguiente pago TOTAL sera de  ");
            System.out.println(NextTax);

        } else if ((impuesto < impuestoReal/2) && (impuesto >= impuestoReal/5)){
            penalizacion = LoDebo/10;
            NextTax = 100 + LoDebo + penalizacion;
            System.out.println("Para este mes le ha faltado este monto para completar su pago: ");
            System.out.println(LoDebo);
            System.out.println("Su penalizacion por tanto sera de ");
            System.out.println(penalizacion);
            System.out.println("El siguiente pago TOTAL sera de  ");
            System.out.println(NextTax);

        } else if (impuesto < impuestoReal/2){
            penalizacion = LoDebo/20;
            NextTax = 100 + LoDebo + penalizacion;
            System.out.println("Para este mes le ha faltado este monto para completar su pago: ");
            System.out.println(LoDebo);
            System.out.println("Su penalizacion por tanto sera de ");
            System.out.println(penalizacion);
            System.out.println("El siguiente pago TOTAL sera de  ");
            System.out.println(NextTax);

        } else {
            System.out.println("Whoa, gracias por tu generosa cooperacion , hermano/a :D ");
        }

        System.out.println("-------------------------------------");
        //-------------------------------------------------------------------------------------------------------------------------------------
        // EJERCICIO 4

        Scanner Scan5 = new Scanner(System.in);
        Scanner Scan6 = new Scanner(System.in);
        double Respuesta1, Respuesta2;
        double A_Years, B_Salario;
        double C_Random = Math.floor(Math.random()*(31)+20);
        System.out.println(C_Random);
        System.out.println("Cual es tu edad");
        A_Years = Scan5.nextFloat();
        System.out.println("Cuanto fue su ultimo salario?");
        B_Salario = Scan6.nextFloat();

        Double Discriminante = Math.pow(B_Salario, 2) - 4*A_Years*C_Random;
        if (Discriminante < 0) {
            System.out.println("La ecuacion no tiene solucion");
            System.out.println("Vuelve a correr el programa");
        } else{
            Respuesta1 = (- B_Salario + Math.sqrt(Discriminante)) / (2*A_Years);
            Respuesta2 = (- B_Salario - Math.sqrt(Discriminante)) / (2*A_Years);
            if (Respuesta1 >= Respuesta2){
                System.out.println("Su tiempo de recracion de hoy es " + Math.abs(Respuesta1) + " Horas");
            } else if (Respuesta2 > Respuesta1){
                System.out.println("Su tiempo de recracion de hoy es " + Math.abs(Respuesta2) + " Horas");
            } else {
                System.out.println("Error");
            }
        }

        System.out.println("-------------------------------------");
        //-------------------------------------------------------------------------------------------------------------------------------------
        // EJERCICIO 5

        Scanner Scan7 = new Scanner(System.in);
        System.out.println("Oficiales .Cuando violaciones al partido detuvieron ayer? ");
        int Atrapados = Scan7.nextInt();
        int ContadorAtrapados = 0;
        int auxiliarContador;

        while (Atrapados > 0){
            if (Atrapados <= 10 ){
                ContadorAtrapados += Atrapados;
                Atrapados = 0;
            }else {
                auxiliarContador = Atrapados % 10;
                ContadorAtrapados += auxiliarContador;
                Atrapados = Atrapados / 10;
            }
        }

        if ((ContadorAtrapados % 9 == 0)){ // múltiplo de 9 implica múltiplo de 3
            System.out.println("Muy bien Oficiales, tienen derecho a descansar");
        } else {
            System.out.println("Hoy no se les hara su descansito :c ");
        }

        System.out.println("-------------------------------------");
        //-------------------------------------------------------------------------------------------------------------------------------------
        // EJERCICIO 6

        Scanner Scan8 = new Scanner(System.in);
        System.out.println("Por favor, escriba el numero secreto, como un entero positivo");
        int NumeroSecreto = Scan8.nextInt();
        int ram;
        int rom = 0;
        int rumba = 0;

        while (NumeroSecreto > 0){
            ram = NumeroSecreto % 10;
            if (rumba % 2 == 0){
                rom += ram;
            } else {
                rom -= ram;
            }
            NumeroSecreto = NumeroSecreto / 10;
            rumba ++;
        }

        if (rom % 11 == 0 ){
            System.out.println("Va a haber recorde de luz por una horita >:D");
        } else {
            System.out.println("Se salvaron, hoy no hay recorde de luz");
        }

        System.out.println("-------------------------------------");
        //-------------------------------------------------------------------------------------------------------------------------------------
        // EJERCICIO 7

        Scanner Scan9 = new Scanner(System.in);
        System.out.println("Ingrese el numero para determinar el presupuesto a la Cultura");
        int PosicionesC = Scan9.nextInt();

        int ContadorC = 0;

        while (PosicionesC > 1){
            if (PosicionesC % 2 == 0 ){
                System.out.println(PosicionesC);
                PosicionesC = PosicionesC / 2;
                ContadorC ++;
            }else {
                System.out.println(PosicionesC);
                PosicionesC = 3*PosicionesC + 1;
                ContadorC ++;
            }
        }
        System.out.println("1");

        System.out.println("Se llego hasta k = " + ContadorC);

        int AuxiliarC = ContadorC * 1000;
        System.out.println("El presupuesto sera de " + AuxiliarC + " Monedas");
    }
}
```
